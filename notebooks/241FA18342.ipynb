## 14.Now we are seperating the dataset into features(X) and Traget Variable(Y) for model training.

```python
X = df_clean.drop(columns=["status", "sl_no", "salary"], errors="ignore")
y = df_clean["status"]

print("X shape:", X.shape)
print("y shape:", y.shape)
display(X.head())
```

## 15.Here preprocessing and model building is done .The feature columns are seperated into Numerical and categorical 

```python
numeric_features = X.select_dtypes(include=np.number).columns.tolist()
categorical_features = X.select_dtypes(exclude=np.number).columns.tolist()

print("Numerical features:", numeric_features)
print("Categorical features:", categorical_features)
```

## 16.Here we are importing train_test_split for splitting the dataset into Training and Testing

```python
from sklearn.model_selection import train_test_split
```

## 17.It divides the datset into Training and Testing and it will maintain same class distribution 
of target variable

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:", X_train.shape)
print("Test set:", X_test.shape)
print("\nTraining target distribution:")
display(y_train.value_counts(normalize=True).mul(100).round(2))
print("\nTest target distribution:")
display(y_test.value_counts(normalize=True).mul(100).round(2))
```

## 18.Here we are importing the preprocessing tools from scikit-learn library.

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

## 19.The below code combines numerical and categorical features into single preprocessor.

```python
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ],
    remainder="drop"
)

print("Preprocessor created successfully.")
```

## 20.Here we are using Logistic Regression.it is a classification algorithm that predicts the probability of a student being placed

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
```

## 21.In the below code we are evaluating the model using Perfomance metrics like Accuracy,Precision,Recall and F1 Score

```python
logistic_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(max_iter=2000, random_state=42))
])

logistic_pipeline.fit(X_train, y_train)

logistic_predictions = logistic_pipeline.predict(X_test)
logistic_probability = logistic_pipeline.predict_proba(X_test)[
    :, list(logistic_pipeline.named_steps["model"].classes_).index("Placed")
]

print("===== Logistic Regression =====")
print("Accuracy :", round(accuracy_score(y_test, logistic_predictions), 4))
print("Precision:", round(precision_score(y_test, logistic_predictions, pos_label="Placed", zero_division=0), 4))
print("Recall   :", round(recall_score(y_test, logistic_predictions, pos_label="Placed", zero_division=0), 4))
print("F1 Score :", round(f1_score(y_test, logistic_predictions, pos_label="Placed", zero_division=0), 4))
print("ROC-AUC  :", round(roc_auc_score((y_test == "Placed").astype(int), logistic_probability), 4))
```

## 22.Here we are importing KNeighborsClassifier

## 23.Here we are using K Nearest Neighbours(KNN)

```python
knn_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", KNeighborsClassifier())
])

knn_pipeline.fit(X_train, y_train)

knn_predictions = knn_pipeline.predict(X_test)
knn_probability = knn_pipeline.predict_proba(X_test)[
    :, list(knn_pipeline.named_steps["model"].classes_).index("Placed")
]

print("===== K-Nearest Neighbors =====")
print("Accuracy :", round(accuracy_score(y_test, knn_predictions), 4))
print("Precision:", round(precision_score(y_test, knn_predictions, pos_label="Placed", zero_division=0), 4))
print("Recall   :", round(recall_score(y_test, knn_predictions, pos_label="Placed", zero_division=0), 4))
print("F1 Score :", round(f1_score(y_test, knn_predictions, pos_label="Placed", zero_division=0), 4))
print("ROC-AUC  :", round(roc_auc_score((y_test == "Placed").astype(int), knn_probability), 4))
```

## 25.we are evaluating the model using Decision Tree

```python
decision_tree_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", DecisionTreeClassifier(random_state=42))
])

decision_tree_pipeline.fit(X_train, y_train)

decision_tree_predictions = decision_tree_pipeline.predict(X_test)
decision_tree_probability = decision_tree_pipeline.predict_proba(X_test)[
    :, list(decision_tree_pipeline.named_steps["model"].classes_).index("Placed")
]

print("===== Decision Tree =====")
print("Accuracy :", round(accuracy_score(y_test, decision_tree_predictions), 4))
print("Precision:", round(precision_score(y_test, decision_tree_predictions, pos_label="Placed", zero_division=0), 4))
print("Recall   :", round(recall_score(y_test, decision_tree_predictions, pos_label="Placed", zero_division=0), 4))
print("F1 Score :", round(f1_score(y_test, decision_tree_predictions, pos_label="Placed", zero_division=0), 4))
print("ROC-AUC  :", round(roc_auc_score((y_test == "Placed").astype(int), decision_tree_probability), 4))
```

## 26.Importing Random Forest

Evaluating the model using RandomForest

```python
random_forest_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(n_estimators=300, random_state=42))
])

random_forest_pipeline.fit(X_train, y_train)

random_forest_predictions = random_forest_pipeline.predict(X_test)
random_forest_probability = random_forest_pipeline.predict_proba(X_test)[
    :, list(random_forest_pipeline.named_steps["model"].classes_).index("Placed")
]

print("===== Random Forest =====")
print("Accuracy :", round(accuracy_score(y_test, random_forest_predictions), 4))
print("Precision:", round(precision_score(y_test, random_forest_predictions, pos_label="Placed", zero_division=0), 4))
print("Recall   :", round(recall_score(y_test, random_forest_predictions, pos_label="Placed", zero_division=0), 4))
print("F1 Score :", round(f1_score(y_test, random_forest_predictions, pos_label="Placed", zero_division=0), 4))
print("ROC-AUC  :", round(roc_auc_score((y_test == "Placed").astype(int), random_forest_probability), 4))
```

## 29.We are storing the trained models in a python dictionary for easy accessing

```python
pipelines = {
    "Logistic Regression": logistic_pipeline,
    "KNN": knn_pipeline,
    "Decision Tree": decision_tree_pipeline,
    "Random Forest": random_forest_pipeline,
    "SVM": svm_pipeline
}

print("Number of trained classification models:", len(pipelines))
```

## 30.Here we are calculating the perfomance metrics and display results 

```python
results = []

for name, pipe in pipelines.items():
    pred = pipe.predict(X_test)
    prob = pipe.predict_proba(X_test)[
        :, list(pipe.named_steps["model"].classes_).index("Placed")
    ]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, pred),
        "Precision": precision_score(y_test, pred, pos_label="Placed", zero_division=0),
        "Recall": recall_score(y_test, pred, pos_label="Placed", zero_division=0),
        "F1": f1_score(y_test, pred, pos_label="Placed", zero_division=0),
        "ROC_AUC": roc_auc_score((y_test == "Placed").astype(int), prob)
    })

results_df = pd.DataFrame(results).sort_values("F1", ascending=False)

display(results_df.round(4))
```

## 31.It uses GridsearchCV to find the best hyperparametres by using cross-validation

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold
rf_pipe = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(random_state=42))
])

rf_params = {
    "model__n_estimators": [200, 400],
    "model__max_depth": [None, 5, 10],
    "model__min_samples_split": [2, 5],
    "model__min_samples_leaf": [1, 2]
}
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

rf_grid = GridSearchCV(
    rf_pipe,
    rf_params,
    cv=cv,
    scoring="f1",
    n_jobs=-1,
    refit=True
)

rf_grid.fit(X_train, y_train)

print("Best RF parameters:", rf_grid.best_params_)
print("Best RF CV F1:", rf_grid.best_score_)
```

```python
lr_pipe = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(random_state=42, max_iter=2000))
])

lr_params = {
    "model__C": [0.1, 1, 10],
    "model__solver": ["lbfgs", "liblinear"],
    "model__max_iter": [1000, 2000]
}

lr_grid = GridSearchCV(
    lr_pipe,
    lr_params,
    cv=cv,
    scoring="f1",
    n_jobs=-1,
    refit=True
)

lr_grid.fit(X_train, y_train)

print("Best LR parameters:", lr_grid.best_params_)
print("Best LR CV F1:", lr_grid.best_score_)
```

## 32.It evaluates the tuned random forest and tuned logistic regression and display the results

```python
tuned_models = {
    "Tuned Random Forest": rf_grid.best_estimator_,
    "Tuned Logistic Regression": lr_grid.best_estimator_
}

tuned_results = []

for name, pipe in tuned_models.items():
    pred = pipe.predict(X_test)
    prob = pipe.predict_proba(X_test)[:, list(pipe.named_steps["model"].classes_).index("Placed")]
    tuned_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, pred),
        "Precision": precision_score(y_test, pred, pos_label="Placed", zero_division=0),
        "Recall": recall_score(y_test, pred, pos_label="Placed", zero_division=0),
        "F1": f1_score(y_test, pred, pos_label="Placed", zero_division=0),
        "ROC_AUC": roc_auc_score((y_test == "Placed").astype(int), prob)
    })

tuned_results_df = pd.DataFrame(tuned_results).sort_values("F1", ascending=False)
display(tuned_results_df.round(4))
```

## 33.It combines original models and tuned models and selects best model as final model

```python
all_final_results = pd.concat([
    results_df,
    tuned_results_df
], ignore_index=True).sort_values("F1", ascending=False)

display(all_final_results.round(4))

best_tuned_name = tuned_results_df.iloc[0]["Model"]
final_model = tuned_models[best_tuned_name]

print("Selected final model:", best_tuned_name)
```

## 34.It identifies the most important features used by the final selected model

```python
if 'final_model' not in locals():
    print("Error: final_model is not defined.")
    print("Please run the model selection cell first (the cell that selects the best tuned model).")
else:
    final_preprocessor = final_model.named_steps["preprocessor"]
    feature_names = final_preprocessor.get_feature_names_out()

    model_obj = final_model.named_steps["model"]

    if hasattr(model_obj, "feature_importances_"):
        importance = model_obj.feature_importances_
        importance_df = pd.DataFrame({
            "Feature": feature_names,
            "Importance": importance
        }).sort_values("Importance", ascending=False)
    elif hasattr(model_obj, "coef_"):
        importance = np.abs(model_obj.coef_[0])
        importance_df = pd.DataFrame({
            "Feature": feature_names,
            "Importance": importance
        }).sort_values("Importance", ascending=False)
    else:
        importance_df = pd.DataFrame()

    if not importance_df.empty:
        display(importance_df.head(20))

        top = importance_df.head(15).sort_values("Importance")
        plt.figure(figsize=(8, 6))
        plt.barh(top["Feature"], top["Importance"])
        plt.xlabel("Importance")
        plt.title("Top Predictive Features")
        plt.tight_layout()
        plt.show()
```

## 37.we are importing mean absolute error ,mean squared error and r2 score to calculate the eroor

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
```

## 38. Here we are comparing the Linear regression and Random Forest to predict the salary for the placed students.

```python
salary_df = df_clean.dropna(subset=["salary"]).copy()
X_salary = salary_df.drop(columns=["salary", "status", "sl_no"], errors="ignore")
y_salary = salary_df["salary"]

salary_num = X_salary.select_dtypes(include=np.number).columns.tolist()
salary_cat = X_salary.select_dtypes(exclude=np.number).columns.tolist()

salary_preprocessor = ColumnTransformer(
    transformers=[
        ("num", Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler())
        ]), salary_num),
        ("cat", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
        ]), salary_cat)
    ]
)
Xtr, Xte, ytr, yte = train_test_split(
    X_salary, y_salary, test_size=0.20, random_state=42
)
salary_models = {
    "Linear Regression": LinearRegression(),
    "Random Forest Regressor": RandomForestRegressor(
        n_estimators=300, random_state=42
    )
}
salary_results = []

for name, model in salary_models.items():
    pipe = Pipeline([
        ("preprocessor", salary_preprocessor),
        ("model", model)
    ])
    pipe.fit(Xtr, ytr)
    pred = pipe.predict(Xte)

    rmse = np.sqrt(mean_squared_error(yte, pred))
    salary_results.append({
        "Model": name,
        "MAE": mean_absolute_error(yte, pred),
        "RMSE": rmse,
        "R2": r2_score(yte, pred)
    })
salary_results_df = pd.DataFrame(salary_results).sort_values("RMSE")
display(salary_results_df.round(2))
```
